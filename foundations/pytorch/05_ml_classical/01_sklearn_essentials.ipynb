```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.pipeline import Pipeline
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.ensemble import RandomForestClassifier
from xgboost import XGBClassifier
```

## 1. Preprocessing Pipeline

Instead of manual encoding, use a structured pipeline to prevent data leakage between train/test splits.

| Component | Description | Usage |
| :--- | :--- | :--- |
| `StandardScaler` | Zero-mean, unit-variance scaling | Essential for linear models/neural nets |
| `OneHotEncoder` | Binary encoding of Categories | Handling categorical features |
| `SimpleImputer` | Filling missing values | Robustness to incomplete data |

```python
num_features = [0, 1, 2] # Index of numerical columns
cat_features = [3, 4]    # Index of categorical columns

preprocessor = ColumnTransformer(
    transformers=[
        ('num', StandardScaler(), num_features),
        ('cat', OneHotEncoder(handle_unknown='ignore'), cat_features)
    ]
)

clf = Pipeline(steps=[
    ('preprocessor', preprocessor),
    ('classifier', RandomForestClassifier())
])
```

## 2. Model Selection (GML vs DL)

✅ **Use when**: Tabular data, structured records, or small datasets (< 100k rows).
❌ **Don't use when**: Data is unstructured (Images, Text, Audio).

### Common Pitfalls
- **Data Leakage**: Fitting an encoder on the ENTIRE dataset before splitting. ALWAYS fit only on `X_train`.
- **Categorical Cardinality**: One-hot encoding a feature with 10k unique values will create a massive, sparse matrix. Use `TargetEncoding` or `Entity Embeddings` (in PyTorch) instead.
- **Scale Sensitivity**: Decision trees (Random Forest, XGBoost) are NOT sensitive to scaling, but MLPs (Neural Nets) are. Standardize if comparing the two.

### Key Takeaways
- `ColumnTransformer` allows different preprocessing logic for different column subsets.
- `Random Forest` is robust to outliers and requires minimal tuning.
- `XGBoost` generally provides higher performance but is more sensitive to hyperparameters.
